# Regresión logística + TF-IDF

## Funcionamiento
Combina pesos TF-IDF del título con variables tabulares en una frontera lineal.

## Ventajas y desventajas
- **Ventaja:** Rápida, inspeccionable y fuerte en texto corto.
- **Desventaja:** No captura interacciones complejas ni contexto semántico.
