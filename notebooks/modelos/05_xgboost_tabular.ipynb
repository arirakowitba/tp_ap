# XGBoost tabular

## Funcionamiento
Boosting secuencial de árboles que corrigen errores previos con regularización.

## Ventajas y desventajas
- **Ventaja:** Mejor de la comparación tabular inicial.
- **Desventaja:** No procesa texto completo y requiere controlar complejidad.
